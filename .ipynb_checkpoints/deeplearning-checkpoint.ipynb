## 1. The Guess (The "Model")

In deep learning, a "model" is just a multiplier. We start with a random Weight (our multiplier).

```python
# Our "Brain"
weight = 0.5 

def predict(input_data):
        return input_data * weight
    
    # Let's try to predict the outcome for the number 10
print(f"Initial Guess: {predict(10)}") # Result: 5.0
```

## 2. The Feedback (The "Loss")

To learn, the AI needs to know the Target (the right answer) and the Error (how far off it was).

```python
target = 20.0  # We want 10 * weight to equal 20
input_val = 10.0

prediction = predict(input_val)
error = target - prediction

print(f"We were off by: {error}") # Result: 15.0
```

## 3. The Correction (The "Learning")

This is the "magic" step. If the error is positive, the weight is too small. If the error is negative, the weight is too big. We nudge the weight slightly using a Learning Rate.

```python
weight = 0.5
learning_rate = .01

# The Training Loop (Repeating the process 100 times)
for i in range(76):
        prediction = input_val * weight
        error = target - prediction
          
      # Nudge the weight in the right direction
        weight = weight + (error * learning_rate)
                        
print(f"Final Weight: {weight:.2f}")
print(f"New Prediction for 10: {input_val * weight:.2f}") 
# Result will be very close to 20!
```

###### Summary:
Forward Pass: You gave the computer an input and it made a guess.

Loss Calculation: You told it how wrong it was.

Backpropagation (The Nudge): It adjusted its internal "multiplier" (weight) so it would be less wrong next time

```python
# 1. Starting Point (The "Wrong" Brain)
w1 = 0.5
w2 = 0.8
lr = 0.01 # Learning Rate (how big of a nudge we take)

def predict(x):
    # Layer 1 -> ReLU -> Layer 2
    layer1 = max(0, x * w1) 
    return layer1 * w2
            
print(f"Before Training (Input 10): {predict(10)}") # This gives 4.0
            
# 2. The Training (The "Learning" Phase)
target = 20.0
input_val = 10.0
            
for epoch in range(300):
   # Forward Pass
   l1_raw = input_val * w1
   l1_act = max(0, l1_raw)
   prediction = l1_act * w2
               
   # Calculate how far off we are
   error = target - prediction
                                                
   # Update Weights (The "Nudge")
   # If error is positive, we increase the weights
   if l1_raw > 0:
      w2 += error * l1_act * lr
      w1 += error * w2 * lr
# 3. The Result (The "Smart" Brain)
print(f"After Training (Input 10): {predict(10):.2f}")  # Should be ~20.0
print(f"After Training (Input -5): {predict(-5):.2f}")  # Should be 0.0
```
